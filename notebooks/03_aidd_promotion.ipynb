# AIDD Promotion Workflow

Purpose: Inspect the AIDM promotion manifest and render the promoted feature module.

Assumption: artifacts/demo/promotion_manifest.json was produced by a successful AIDM workflow with decision promote.

```
import json
from pathlib import Path
from power_forecasting.cli import run_aidd_workflow
```

```
artifact_root = Path("artifacts/demo")
manifest_path = artifact_root / "promotion_manifest.json"
```

```
manifest = json.loads(manifest_path.read_text(encoding="utf-8"))
generated_module = run_aidd_workflow(artifact_root, manifest=manifest_path)
manifest, generated_module
```
